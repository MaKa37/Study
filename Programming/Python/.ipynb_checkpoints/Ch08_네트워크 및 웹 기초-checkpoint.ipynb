# 1. HTTP 프로토콜(Requests)

```python
import requests
import json

# 학습용 무료 REST API 엔드포인트
# (참고: 실제로 서버의 데이터가 영구적으로 바뀌지는 않지만, 성공/실패 상태 코드와 결과 포맷을 정확히 반환해 줍니다.)
BASE_URL = "https://jsonplaceholder.typicode.com/posts"

def print_result(method_name, response):
    """가독성을 높이기 위한 결과 출력용 헬퍼 함수"""
    print(f"[{method_name}] 상태 코드: {response.status_code}")
    # 응답 본문이 너무 길지 않게 들여쓰기하여 출력
    print(json.dumps(response.json(), indent=2, ensure_ascii=False))
    print("-" * 50)

# =====================================================================
# 1. GET (조회)
# 속성: 안전성(O), 멱등성(O)
# 용도: 데이터를 읽어올 때 사용. URL에 모든 정보가 포함됨.
# =====================================================================
# 1번 게시글 조회 요청
response_get = requests.get(f"{BASE_URL}/1")
print_result("GET", response_get)

# =====================================================================
# 2. POST (생성)
# 속성: 안전성(X), 멱등성(X)
# 용도: 새로운 데이터를 등록. 데이터는 URL이 아닌 본문(Body)에 담아 전송.
# =====================================================================
new_post_data = {
    "title": "크롤링 학습하기",
    "body": "HTTP 메서드 이해하기",
    "userId": 1
}
# requests.post의 'json' 파라미터를 사용하면 자동으로 헤더에 'Content-Type: application/json'이 추가됩니다.
response_post = requests.post(BASE_URL, json=new_post_data)
print_result("POST", response_post) # 성공 시 201 (Created) 상태 코드 반환

# =====================================================================
# 3. PUT (전체 수정)
# 속성: 안전성(X), 멱등성(O)
# 용도: 기존 자원을 완전히 새로운 데이터로 덮어쓰기. (여러 번 보내도 결과 동일)
# =====================================================================
put_update_data = {
    "id": 1,
    "title": "수정된 제목 (PUT)",
    "body": "내용이 완전히 교체되었습니다.",
    "userId": 1
}

# 1번 게시글 전체를 덮어씀
response_put = requests.put(f"{BASE_URL}/1", json=put_update_data)
print_result("PUT", response_put)

# =====================================================================
# 4. PATCH (일부 수정)
# 속성: 안전성(X), 멱등성(X/O)
# 용도: 자원의 일부분만 변경. (여기서는 title만 변경하므로 멱등성 유지)
# =====================================================================
patch_update_data = {
    "title": "제목만 살짝 수정(PATCH)"
    # body와 userId는 보내지 않으므로 기존 데이터가 유지됨
}

# 1번 게시글의 제목만 수정
response_patch = requests.patch(f"{BASE_URL}/1", json=patch_update_data)
print_result("PATCH", response_patch)

# =====================================================================
# 5. DELETE (삭제)
# 속성: 안전성(X), 멱등성(O)
# 용도: 특정 자원을 삭제. (이미 삭제된 것을 또 삭제 요청해도 상태는 '삭제됨'으로 동일)
# =====================================================================
# 1번 게시글 삭제 요청 (Body가 필요 없음)
response_delete = requests.delete(f"{BASE_URL}/1")
print_result("DELETE", response_delete)
```

```python
# GET, POST, PUT, UPDATE, PATCH

r_get = requests.get(f"{BASE_URL}/1")
print_result("GET", r_get)

r_post_dict = {
    "userId": 1,
    "id": 1, # Post Number -> 100번까지 포스트가 있어서 101로 반환됨
    "title": "Learning Python API",
    "body": "HTTP Protocol"
}
r_post = requests.post(f"{BASE_URL}", json=r_post_dict)
print_result("POST", r_post)

r_put_dict = {
    "id": 1,
    "body": "HTTP Protocol v2"
}
r_put = requests.put(f"{BASE_URL}/1", json=r_put_dict)
print_result("PUT", r_put)

r_patch_dict = {
    "title": "Learning Python HTTP Protocol"
}

r_patch = requests.patch(f"{BASE_URL}/1", json=r_patch_dict)
print_result("PATCH", r_patch)

r_delete = requests.delete(f"{BASE_URL}/1")
print_result("DELETE", r_delete)
```

```python
from fastapi import FastAPI
from fastapi.testclient import TestClient

# =====================================================================
# 1. 서버 측 구현 (RESTful API 라우팅 설계)
# =====================================================================
app = FastAPI()

# 가상의 데이터베이스 세팅
MOCK_USERS = [
    {"id": 123, "name": "김철수", "role": "admin"},
    {"id": 124, "name": "이영희", "role": "user"},
]
MOCK_ORDERS = [
    {"id": 45, "user_id": 123, "item": "노트북"},
    {"id": 46, "user_id": 124, "item": "무선 마우스"},
]

# [1] 경로 변수 (Path Variable)
@app.get("/users/{user_id}")
def get_user(user_id: int):
    # 전달받은 user_id로 자원 식별
    user = next((u for u in MOCK_USERS if u["id"] == user_id), None)
    return {"status": "success", "data": user}

# [2] 계층적 경로 (Hierarchical Path)
@app.get("/users/{user_id}/orders/{order_id}")
def get_user_order(user_id: int, order_id: int):
    # 포함 관계(특정 유저의 특정 주문) 자원 식별
    order = next((o for o in MOCK_ORDERS if o["user_id"] == user_id and o["id"] == order_id), None)
    return {"status": "success", "data": order}

# [3 & 4] 쿼리 파라미터, 페이징 및 복합 쿼리 (Query Parameter, Pagination)
@app.get("/users")
def get_users(role: str = None, page: int = 1, limit: int = 10):
    # 1. 쿼리 파라미터 필터링
    filtered_users = MOCK_USERS
    if role:
        filtered_users = [u for u in filtered_users if u["role"] == role]
    
    # 2. 페이징 로직 (테스트를 위해 가상 데이터를 임의로 복제하여 양을 늘림)
    extended_users = filtered_users * 50  
    
    start_idx = (page - 1) * limit
    end_idx = start_idx + limit
    paginated_users = extended_users[start_idx:end_idx]

    return {
        "status": "success",
        "meta": {"role_filter": role, "page": page, "limit": limit, "returned_count": len(paginated_users)},
        "data": paginated_users
    }


# =====================================================================
# 2. 클라이언트 측 호출 (requests 구현 시뮬레이션)
# =====================================================================
# TestClient는 파이썬 내부에서 requests 라이브러리와 100% 동일한 인터페이스를 제공합니다.
client = TestClient(app)

def run_simulation():
    print("--- [1] 경로 변수 (Path Variable) ---")
    user_id = 123
    response1 = client.get(f"/users/{user_id}")
    print(f"생성된 URL : {response1.request.url}")
    print(f"응답 결과  : {response1.json()}\n")

    print("--- [2] 계층적 경로 (Hierarchical Path) ---")
    order_id = 45
    response2 = client.get(f"/users/{user_id}/orders/{order_id}")
    print(f"생성된 URL : {response2.request.url}")
    print(f"응답 결과  : {response2.json()}\n")

    print("--- [3] 쿼리 파라미터 (Query Parameter) ---")
    response3 = client.get("/users", params={"role": "admin"})
    print(f"생성된 URL : {response3.request.url}")
    print(f"응답 결과  : {response3.json()['meta']}\n") # 출력 길이를 위해 meta만 출력

    print("--- [4] 페이징 및 복합 쿼리 (Pagination & Complex) ---")
    params = {"role": "admin", "page": 2, "limit": 3}
    response4 = client.get("/users", params=params)
    print(f"생성된 URL : {response4.request.url}")
    print(f"응답 메타  : {response4.json()['meta']}")
    print(f"응답 데이터: {response4.json()['data']}")

if __name__ == "__main__":
    run_simulation()
```
