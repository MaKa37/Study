# 1. Exception Handling

```python
def calculate_division(num1, num2):
    print(f"\n--- [실행] {num1} / {num2} 연산 시작 ---")

    try:
        # 1. [Try] 예외 발생 가능성이 있는 메인 로직
        # 의도적으로 형변환(float)과 나눗셈(/)을 수행하여 에러 유발 지점 생성
        value1 = float(num1)
        value2 = float(num2)
        result = value1 / value2

    except ValueError as ve:
        # 2. [Except] 특정 예외 처리 1: 문자를 숫자로 바꿀 수 없을 때
        print(f"  ❌ [ValueError] 숫자로 변환할 수 없는 값이 있습니다. (상세: {ve})")

    except ZeroDivisionError as ze:
        # 3. [Except] 특정 예외 처리 2: 0으로 나눌 때
        print(f"  ❌ [ZeroDivisionError] 수학적으로 0으로 나눌 수 없습니다. (상세: {ze})")

    except Exception as e:
        # 4. [Except] 포괄적 예외 처리 (반드시 구체적인 예외들보다 아래에 작성)
        print(f"  ❌ [알 수 없는 에러] 예기치 못한 문제가 발생했습니다: {e}")

    else:
        # 5. [Else] 에러가 단 하나도 발생하지 않았을 때만 실행
        # 로직이 무사히 통과했을 때의 결과 출력이나 후속 작업을 이곳에 작성합니다.
        print(f"  ✅ [정상 처리] 연산 성공! 결과 = {result:.2f}")

    finally:
        # 6. [Finally] 성공/실패, 심지어 함수가 return으로 종료되어도 무조건 실행
        # 주로 열어둔 파일(f.close())이나 DB 연결 해제 등 '마무리 작업'을 담당합니다.
        print("  🏁 [종료] 할당된 자원 해제 및 시스템 로그 기록 완료")

# ==========================================
# 테스트 케이스 실행 (결과 확인용)
# ==========================================
if __name__ == "__main__":
    # 케이스 1: 완벽하게 정상적인 입력 (try -> else -> finally 흐름)
    calculate_division(10, 2)
    
    # 케이스 2: 0으로 나누는 수학적 오류 (try -> ZeroDivisionError -> finally 흐름)
    calculate_division(10, 0)
    
    # 케이스 3: 숫자가 아닌 문자열 입력 (try -> ValueError -> finally 흐름)
    calculate_division("파이썬", 5)
```

| **파일 실행 방식**                       | **내부 __name__ 변수의 값**  | **if문 조건 충족** | **결과**           |
| ---------------------------------- | ---------------------- | ------------- | ---------------- |
| **직접 실행** (예: `python my_code.py`) | `"__main__"`           | **`True`**    | 블록 내부 코드 **실행됨** |
| **모듈로 임포트** (예: `import my_code`)  | 파일 이름 (예: `"my_code"`) | `False`       | 블록 내부 코드 **무시됨** |

# 2. Text File I/O

```python
import os

def run_file_io_tutorial(filename):
    print(f"=== '{filename}' 입출력 테스트 시작 ===\n")

    # 1. 파일 쓰기 ('w' 모드) - 기존 파일이 있으면 덮어씁니다.
    print("[1단계] 파일 쓰기 (w)")
    try:
        # with 구문을 사용하면 작업 종료 시 f.close()가 자동 호출됩니다.
        with open(filename, 'w', encoding='utf-8') as f:
            f.write("첫 번째 줄: write()로 문자열을 기록합니다.\n")
            f.writelines([
                "두 번째 줄: writelines()로 리스트를 기록합니다.\n",
                "세 번째 줄: 줄 바꿈 문자(\\n)를 직접 제어해야 합니다.\n"
            ])
        print("  ✅ 파일 생성 및 데이터 쓰기 성공\n")
        
    except Exception as e:
        print(f"  ❌ 파일 쓰기 실패: {e}\n")

    # 2. 파일 추가 ('a' 모드) - 기존 파일의 맨 끝에 내용을 이어 붙입니다.
    print("[2단계] 파일 추가 (a)")
    try:
        with open(filename, 'a', encoding='utf-8') as f:
            f.write("네 번째 줄: 'a' 모드로 추가된 내용입니다.\n")
        print("  ✅ 기존 파일에 데이터 추가 성공\n")

    except Exception as e:
        print(f"  ❌ 파일 추가 실패: {e}\n")

    # 3. 파일 읽기 ('r' 모드) - 상황에 맞는 세 가지 읽기 방식 비교
    print("[3단계] 파일 읽기 (r)")
    try:
        # 방법 A: 파일 전체를 하나의 문자열로 읽기 (파일이 작을 때 유리)
        print("  --- [방법 A] f.read() ---")
        with open(filename, 'r', encoding='utf-8') as f:
            content = f.read()
            print(content, end="")

        # 방법 B: 모든 줄을 읽어 리스트로 반환 (데이터를 메모리에 쌓아두고 가공할 때)
        print("\n  --- [방법 B] f.readlines() ---")
        with open(filename, 'r', encoding='utf-8') as f:
            lines = f.readlines()
            print(f" 총 {len(lines)}줄의 데이터를 리스트로 읽어왔습니다.")

        # 방법 C: 파일 객체를 이터레이터로 사용하여 한 줄씩 순회 (대용량 파일에 필수)
        print("\n  --- [방법 C] for 반복문 (가장 권장됨) ---")
        with open(filename, 'r', encoding='utf-8') as f:
            for idx, line in enumerate(f, start=1):
                # 파일에 저장된 \n과 print의 기본 \n이 중복되므로 strip()으로 제거
                print(f"  {idx}행: {line.strip()}")

    except FileNotFoundError:
        print(f"  ❌ 에러: '{filename}' 파일을 찾을 수 없습니다.")

    except Exception as e:
        print(f"  ❌ 알 수 없는 에러 발생: {e}")

    # 4. 마무리 자원 정리 (테스트 파일 삭제)
    print("\n[4단계] 테스트 종료 및 파일 정리")
    if os.path.exists(filename):
        os.remove(filename)
        print("  🏁 생성된 테스트 파일을 시스템에서 완전히 삭제했습니다.")

# ==========================================
# 실행부
# ==========================================
if __name__ == "__main__":
    test_file_name = "tutorial_sample.txt"
    run_file_io_tutorial(test_file_name)
```

# 3. JSON 직렬화 및 역직렬화 (JSON Serialization)

```python
import json

# 1. Python 객체를 JSON 문자열로 변환 (dumps)
python_data = {
    "name": "홍길동",
    "age": 30,
    "is_student": True,
    "hobbies": ["독서", "코딩"]
}

# ensure_ascii=False: 한글 깨짐 방지 / indent=4: 보기 좋게 들여쓰기
json_string = json.dumps(python_data, ensure_ascii=False, indent=4)
print("--- Python Dict 객체 ---")
print(python_data)

print("--- JSON 문자열 ---")
print(json_string)
print("-------------------------------------------------")

# 2. JSON 문자열을 Python 객체로 변환 (loads)
parsed_data = json.loads(json_string)
print("\n--- Python 객체 ---")
print(parsed_data["name"])
print(parsed_data)
print("-------------------------------------------------")
# 3. JSON 파일 입출력 (dump, load)
# 파일에 쓰기
with open("data.json", "w", encoding="utf-8") as f:
    json.dump(python_data, f, ensure_ascii=False, indent=4)

with open("data.json", "r", encoding="utf-8") as f:
    loaded_data = json.load(f)
    print(f"읽어온 데이터: {loaded_data}")

if os.path.exists("data.json"):
    os.remove("data.json")
    print("  🏁 생성된 테스트 파일을 시스템에서 완전히 삭제했습니다.")
```
