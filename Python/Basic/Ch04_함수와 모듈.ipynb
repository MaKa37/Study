# 1. 함수

## 1.1 함수의 기본 구조(정의, 호출, 매개변수, 반환값)

```python
# 1. 함수 정의(타입 힌트와 독 스트링을 포함한 완성형 구조)
def add_numbers(a: int, b: int = 10) -> int:
    """
    두 정수를 입력받아 합계를 반환하는 함수입니다.

    :param a: 첫 번째 정수 (위치 매개변수)
    :param b: 두 번째 정수 (기본 매개변수, 기본 값 10)
    :return : 두 정수의 합
    """

    result = a + b # 함수 본문(들여쓰기 규칙 적용)
    return result  # 결과값 반환

print(add_numbers(10))     # 20
print(add_numbers(10, 20)) # 30
print("-------------------------------------------------")
help(add_numbers)
print("-------------------------------------------------")
```

```python
def calculate_total_sum(*args):
    """
    - *args: Arguments(인자들)
    - 입력받은 모든 숫자의 합을 계산하는 함수입니다.
    """

    # 1. args의 자료형과 내부 데이터 확인
    print(f"[Data Type] args: {type(args)}")
    print(f"[Input Data] args: (args)")

    # 2. Tuple 이므로 for문을 통해 순회 가능합니다.
    total = 0
    for num in args: # *args / Ex: Input Data(10, 20 ,30) 
        total += num # total: 60

    return total

# === 호출 예시 === #
print(calculate_total_sum(10, 20, 30))
```

```python
# 가변 키워드 매개변수
def print_user_profile(**kwargs):
    """
    - **kwargs: Keyword Arguments(단어 인자들)
    """

    # 1. kwargs의 자료형과 내부 데이터 확인
    print(f"[Data Type] kwargs: {type(kwargs)}")
    print(f"입력된 데이터 kwargs: {kwargs}")

    # 2. Dictionary 이므로 .item()를 통해 키(key)와 값(value) 추출 가능
    print("-------------------------------------------------")
    for k, v in kwargs.items():
        print(f"Key: {k}ㅣValue: {v}")

# === 호출 예시 === #
print_user_profile(name="홍길동", age=25)
print("-------------------------------------------------")
print_user_profile(name="김철수", age=30, job="개발자", location="경기도")
```

## 1.2 변수의 유효 범위(지역 변수와 전역 변수)

```python
# Variable Scope(변수 유효범위)

## 지역 변수(Local Variable)
def check_local_scope():
    local_msg = "이 변수는 함수 안에서만 유효합니다."
    print(f"[함수 내부] {local_msg}")

check_local_scope()
# print(local_msg) / NameError: name 'local_msg' is not defined
print("-------------------------------------------------")

## 전역 변수(Global Variable)
global_msg = "이 변수는 프로그램 전체에서 볼 수 있습니다."

def check_global_scope():
    print(f"[함수 내부] {global_msg}")

check_global_scope()
print(f"[함수 외부] {global_msg}")
print("-------------------------------------------------")
```

```python
# 전역 변수 초기화
user_count = 0

def increase_count_wrong():
    # global 선언 없이 수정을 시도하면 에러 발생
    # user_count = user_count + 1
    # -> UnboundLocalError: cannot access local variable 'user_count' where it is not associated with a value
    pass

def increase_count_correct():
    # '이 함수 안에서 사용할 user_count는 밖에 있는 전역 변수다'라고 선언
    global user_count

    user_count += 1
    print(f"[함수 내부] 방문자 수 증가: {user_count}")

print(f"초기 방문자 수: {user_count}")
increase_count_correct()
increase_count_correct()
print(f"[함수 외부] 최종 방문자 수 {user_count}")
```

```python
# 변수 우선순위(Variable Shadowing)
# Local > Enclosed > Global > Built-in

"""
전역 변수와 지역 변수의 이름이 같을 때, 파이썬이 어떤 변수를 우선순위로 두는지 보여주는 중요한 예시입니다.
파이썬은 항상 자신이 속한 가장 가까운 범위(지역)의 변수를 우선합니다.
"""

# 전역 변수(Global Variable)
player_name = "홍길동 (전역 변수)"

def play_game():
    # 전역 변수와 이름이 똑같은 지역 변수 생성 (global 선언 없음)
    # 이 순간부터 이 함수 안에서는 새로운 'player_name'이 전역 변수를 가려버립니다(Shadowing).
    player_name = "김철수 (지역 변수)"

    print(f"[함수 내부] 현재 플레이어: {player_name}")

print(f"[함수 외부] 게임 시작 전: {player_name}")
play_game()
print(f"[함수 외부] 게임 시작 후: {player_name}")
```
