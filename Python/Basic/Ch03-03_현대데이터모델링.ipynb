## 1.1 데이터 클래스(DataClasses)

```python
from dataclasses import dataclass

# 1. Dataclass를 사용한 데이터 모델링

@dataclass
class Product:
    name: str
    price: int
    discounted: bool = False # 기본 값 설정 가능

# 객체 생성
apple = Product("Apple", 1500)
print(apple) # 출력: Product(name='Apple', price=1500, discounted=False)
print(type(apple)) # 출력: <class '__main__.Product'>

# 동등성 비교 (내부 값이 같으면 같은 객체로 취급)
apple2 = Product("Apple", 1500)
print(apple == apple2) # 출력: True (일반 클래스였다면 메모리 주소가 달라 False가 나옵니다)
```

## 1.2 상수 집합(Enum)

```python
from enum import Enum, auto

# Enum 클래스 정의
class OrderStatus(Enum):
    PENDING = 1     # 대기 중
    PROCESSING = 2  # 처리 중
    SHIPPED = 3     # 배송 중

    # auto() 사용 시 1, 2, 3 숫자를 자동 할당함.
    DELIVERED = auto() # 배송 완료(4)

# Enum
current_status = OrderStatus.PENDING

print(current_status.name)  # 출력: PENDING
print(current_status.value) # 출력: 1
```
