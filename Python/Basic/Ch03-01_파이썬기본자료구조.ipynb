# 1 시퀀스형

## 1.1 List

```python
# List
arr1 = [1, 2, 3]
arr2 = [3, 4, 5]
arr3 = list(range(5))

print("-------------------------------------------------")

print(arr1)
print(arr2)
print(arr3)

print("-------------------------------------------------")

arr1.append(4) # 리스트의 맨 뒤에 숫자 4 추가
print(arr1)

arr1.insert(0, 0) # 맨 앞인 0번째 인덱스에 숫자 0 삽입
print(arr1)

print("-------------------------------------------------")

arr4 = []
print(arr4)
arr4.extend(arr2) # 빈 리스트에 arr2 리스트를 병합
print(arr4)

print("-------------------------------------------------")

print((arr1.pop())) # [0, 1, 2, 3, 4] 리스트에서 맨 뒤에 반환된 값 출력
print(arr1)         # [0, 1, 2, 3] 리스트에서 맨 뒤에 반환된 값이 삭제된 모습

arr1.remove(0) # [0, 1, 2, 3] 리스트에서 순차 조회하여 첫 번째 x값 삭제(x가 여러개 일 경우)
print(arr1)

del arr1[2] # [1, 2, 3] 리스트에서 순차 조회하여 x 인덱스에 해당하는 값 삭제
print(arr1)

print("-------------------------------------------------")

print(arr1.index(2)) # 2번째 인덱스 번호인 1을 반환함.
print(arr1.count(1)) # 리스트 내 값이 1인 요소의 개수를 반환함.

print("-------------------------------------------------")
arr3.reverse() # 리스트 요소 순번 역순으로 뒤집기
print(arr3)

arr3.sort() # 오름차순 / 내림차순 정렬
print(arr3)
```

## 1.2 Tuple

```python
# Tuple
tup1 = (1, 2, 3)
tup2 = tuple(arr2)
tup3 = tuple(range(5))

print(tup1)
print(tup2)
print(tup3)
print("-------------------------------------------------\n")

print(tup1.index(3)) # 숫자 3이 위치한 인덱스 번호 반환
print(tup1.count(3)) # 숫자 3의 개수를 반환

print("-------------------------------------------------\n")

print(tup1 + tup2)
print(tup1 * 2)

print("-------------------------------------------------\n")

print(sorted(tup1 + tup2)) # 튜플을 정렬하면서 리스트로 반환 됨.
tup4 = (1, 2, 3, [4, 5])

print(tup4) # 튜플안에 리스트도 생성 가능하며, 해당 리스트는 가변적임"
```

## 1.3 Str

```python
s1 = 'A'
s2 = "A"
s3 = '''A'''
s4 = """A"""
s5 = str(1000)
s6 = 9999
print(s1, s2, s3, s4, s5, f"{s6}", sep=" ")

print("-------------------------------------------------\n")

s7 = "Hello, Python!"
s7sub = "llo"
print(s7.find(s7sub))  # 부분 문자열이 몇 번째 인덱스 번호에 나오는지 순번 반환
print(s7.index("llo")) # 조회는 인덱스 맨 앞 순번부터 조회한다.
print(s7.count("l"))   # 문자열에서 "l"의 개수 카운트
print(s7.replace("Python!", "World")) # 값을 치환하여 반환하는 메서드 (메모리에 저장 X)
print(s7)
print(s7.upper())
print(s7.lower())
print((s7 + "    ").strip()) # 우측 공백 뮨자 제거
print(s7.split(", "))

print(s7[1:])
print(s7[7:])
print(len(s7))
print("-------------------------------------------------\n")
print(s7.startswith("Hello"))
print(s7.endswith("World"))
print(s7.isdigit())
print(s7.isalpha())
```

## 1.4 Range

```python
# Range

r1 = range(5) # 0, 1, 2, 3, 4
r2 = range(1, 10) # 1, 2, 3, ... 7, 8, 9
r3 = range(1, 10, 2) # 1, 3, 5, 7, 9

print(list(r1))
print(list(r2))
print(list(r3))

print("-------------------------------------------------")

print(r1.index(3))
print(r1.count(3))

print("-------------------------------------------------")

r4 = r1[2:]
print(r4)
print(list(r4))
print(3 in r4)
```

# 2. 매핑형

## 2.1 Dictionary

```python
d1 = {'a': 1, 'b': 2}
d2 = dict(a = 1, b = 2)
d3 = dict([('a', 1), ('b', 2)])

print(d1)
print(d2)
print(d3)
print("-------------------------------------------------")
key_list = ['a', 'b', 'c']
value_list = [1, 2, 3]

d4 = {k: v for k, v in zip(key_list, value_list)}
print(d4)
print(d4['a'], d4['b'], d4['c']) # d4['d']는 Key가 존재하지 않으므로 ValueError 발생
print(d4.keys())
print(d4.values())
print("-------------------------------------------------")
d4['a'] = 99
print(d4['a'])

d4.pop('a', 'default')
print(d4)

d4.popitem() # 마지막으로 삽입된 키와 값 반환 후 삭제(LIFO)
print(d4)

del d4['b']
print(d4)

d4.update(d1)
print(d4)

d4.clear()
print(d4)
```

# 3. 집합형

## 3.1 Set

```python
s1 = {1, 2, 3}
s2 = set(range(1, 3))
s3 = {x for x in s1}
empty_set = set()

print(s1)
print(s2)
print(s3)
print(empty_set)
print(type(s1))
print("-------------------------------------------------")

s1.add(4)
s1.add(4)
s1.add(5)
print(s1) # 중복 값은 추가되지 않음

empty_set.update(s1) # update(iter)
print(empty_set)
print("-------------------------------------------------")
empty_set.remove(5)  # 값이 없을 시 KeyError 발생
empty_set.discard(4)
empty_set.discard(4) # 값이 없어도 Error는 발생하지 않는다.
empty_set.pop()      # FIFO

print(empty_set)
empty_set.clear()
print(empty_set)
print("-------------------------------------------------")

s = { 1, 2, 3 }
t = { 3, 4, 5 }

u = s | t # 합집합 / union(t)
print(u)

i = s & t # 교집합 / intersection(t)
print(i)

d = s - t # 차집합 / difference(t)
print(d)

sym = s ^ t # 대칭차집합 / symmetric_difference(t)
print(sym)

print(s <= t) # 부분집합 체크 T / F / issubset(t)

print(s >= t) # 상위집함 체크 T / F / isuperset(t)

print(s.isdisjoint(t)) # 교집합이 없는지 체크 T / F
```

## 3.2 FrozenSet

```python
fs1 = frozenset([1, 2, 3])
empty_fs = frozenset()

print(fs1)
print(empty_fs)

print("-------------------------------------------------")

# 합, 교, 차집합 용도의 Set
s = { 1, 2, 3 }
t = { 3, 4, 5 }

fs_s = frozenset(s)
fs_t = frozenset(t)

print(fs_s)
print(fs_t)

print("-------------------------------------------------")

fs_i = fs_s | fs_t 
fs_d = fs_s & fs_t
fs_sym = fs_s ^ fs_t

print(fs_i)
print(fs_d)
print(fs_sym)

print("-------------------------------------------------")

print(fs_s <= fs_t)
print(fs_s >= fs_t)
print(fs_s.isdisjoint(fs_t))
```
