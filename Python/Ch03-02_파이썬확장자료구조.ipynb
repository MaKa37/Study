```python
from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])  # 큐 초기화
    order = []

    while queue:
        node = queue.popleft()  # 맨 앞에서 꺼냄 (FIFO)
        order.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)  # 맨 뒤에 추가

    return order

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C']
}

print(bfs(graph, 'A'))  # ['A', 'B', 'C', 'D']
```

# 1. 확장 자료구조

## 1.1 양방향 큐(deque)

```python
from collections import deque

dq1 = deque()
dq2 = deque(range(3))
dq3 = deque(range(3), 3)

print(dq1)
print(dq2)
print(dq3) # 최대 길이가 고정된 deque

print("-------------------------------------------------")

dq2.append(3)
dq2.appendleft(-1)
print(dq2)

dq3.append(3)
dq3.appendleft(-1) # 최대 길이에 대한 제약으로 뒤 요소와 앞의 요소가 빠지고 새로 추가됨.
print(dq3)

dq_alphas = deque(["A", "B", "C"])
dq_nums = deque([1, 2, 3])

dq1.extend(dq_alphas)
print(dq1)

dq1.extendleft(dq_nums) # deque의 맨앞에 역순으로 삽입됨.
print(dq1)

print("-------------------------------------------------")

dq2.pop() # deque([-1, 0, 1, 2, 3]) <-
print(dq2)

dq2.popleft() # deque -> ([-1, 0, 1, 2])
print(dq2)

dq2.remove(1)
print(dq2)

dq2.clear()
print(dq2)

print("-------------------------------------------------")

dq3.rotate(1)
print(dq3)

dq3.rotate(1)
print(dq3)

dq3.rotate(1)
print(dq3)

dq3.rotate(-1)
print(dq3)

print(dq3.count(1))
print(dq3.index(-1))
```

## 1.2 빈도 분석(Counter)

```python
from collections import Counter
c1 = Counter(["A", "B", "B", "C", "C", "C"])
c2 = Counter({'A': 1, 'B': 2, 'C': 3})

print(c1)
print(c2)

print("-------------------------------------------------")

print(c1.most_common(1))
print(c1.most_common(2))
print(c1.most_common(3))

print("-------------------------------------------------")

print(list(c1.elements()))
print(list(c2.elements()))
print(c1.total())

print("-------------------------------------------------")

c1.update(c2)
print(c1)

c1.subtract(c2)
print(c1)

print("-------------------------------------------------")

del c1['A']
print(c1)
```

## 1.3 자동 기본값(defaultdict)

```python
from collections import defaultdict

group_by_int = defaultdict(int)
group_by_list = defaultdict(list)
group_by_set = defaultdict(set)

print(group_by_int)
print(group_by_list)
print(group_by_set)

print("-------------------------------------------------")
group_by_int['Apple'] += 3
group_by_int['Banana'] += 2
group_by_int['WarterMelon'] += 1
print(group_by_int)
print(dict(group_by_int))

print("-------------------------------------------------")
group_by_list['Team'].append('A')
group_by_list['Team'].append('B')
print(group_by_list)
print(dict(group_by_list))

print("-------------------------------------------------")
group_by_set['A'].add(1)
group_by_set['A'].add(1)
group_by_set['A'].add(1)
print(group_by_set)
print(dict(group_by_set))

print("-------------------------------------------------")
print(group_by_int.keys())
print(group_by_int.values())
print(group_by_int.items())

group_by_int.update(group_by_list)
print(group_by_int)
```

# 2. 가독성 및 특수 제어

## 2.1 이름 기반 접근(namedtuple)

```python
from collections import namedtuple

Point1 = namedtuple('Point1', ['x', 'y'])
print(Point1)
print("-------------------------------------------------")

p = Point1(10, 20)
print(p)
print(f"{p.x}, {p.y}")
print(p[0], p[1])
print("-------------------------------------------------")

a_list = [[1, 2], [3, 4], [5, 6]]
p2 = Point1._make(a_list[0])
p3 = Point1._make(a_list[1]) 
p4 = Point1._make(a_list[2])
print([p2, p3, p4])

p_list = [Point1._make(data) for data in a_list]
print(p_list)

d_list = [p._asdict() for p in p_list]
print(d_list)
print("-------------------------------------------------")

print(p._replace(x='A', y='B'))
print(p._fields)
print(p)
```

## 2.2 순서 제어(OrderedDict)

```python
from collections import OrderedDict

od1 = OrderedDict()
od2 = OrderedDict(a = 1, b = 2)
od3 = OrderedDict([('a', 1), ('b', 2)])

print(od1)
print(od2)
print(od3)
print("-------------------------------------------------")

od2.move_to_end('a')
print(od2)

od2.move_to_end('a', last=False)
print(od2)
print("-------------------------------------------------")

print(od2.popitem())
print(od2)

print(od3.popitem(last=False))
print(od3)
print("-------------------------------------------------")

print(od2 == od3) # OrderedDict([('a', 1)]) != OrderedDict([('b', 2)])
print(od2.keys())
print(od2.values())
print(od2.items())

od2.update(od3)
print(od2)
```

## 2.3 계층적 매핑(ChainMap)

```python
from collections import ChainMap

dict1 = {'a': 1, 'b': 2}
dict2 = {'A': 8, 'B': 9}

cm = ChainMap(dict1, dict2)
print(cm)
print(cm.maps)
print(f"cm: {cm.maps}")
print("-------------------------------------------------")

child_cm = cm.new_child()
child_cm['a'] = 999
child_cm['c'] = 100

print(f"child_cm.maps: {child_cm.maps}")
print(f"child_cm['a']: {child_cm['a']}")
print(f"dict(child_cm): {dict(child_cm)}")

print("-------------------------------------------------")
pcm = child_cm.parents
print(pcm)

print(f"pcm['a']: {pcm['a']}")

pcm['a'] = 999
print(f"pcm['a'] = 999: {pcm['a']}")

del pcm['a']
print(dict(pcm))
```

# 3. 커스텀 자료구조

## 3.1 안전한 래퍼(UserDict, UserList, UserString)

```python
from collections import UserDict
from collections import UserList
from collections import UserString

ud = UserDict({'a': 1, 'b': 2})
print(ud.data)

ul = UserList([1, 2, 3, 4, 5])
print(ul.data)

us = UserString("Hello, World!")
print(us.data)

class UpperDict(UserDict):
    def __setitem__(self, key, value):
        super().__setitem__(key.upper(), value)

ud2 = UpperDict(ud)
print(ud2)
```
