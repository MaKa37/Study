# 1. 함수

## 1.1 함수의 기본 구조(정의, 호출, 매개변수, 반환값)

```python
# 1. 함수 정의(타입 힌트와 독 스트링을 포함한 완성형 구조)
def add_numbers(a: int, b: int = 10) -> int:
    """
    두 정수를 입력받아 합계를 반환하는 함수입니다.

    :param a: 첫 번째 정수 (위치 매개변수)
    :param b: 두 번째 정수 (기본 매개변수, 기본 값 10)
    :return : 두 정수의 합
    """

    result = a + b # 함수 본문(들여쓰기 규칙 적용)
    return result  # 결과값 반환

print(add_numbers(10))     # 20
print(add_numbers(10, 20)) # 30
```
