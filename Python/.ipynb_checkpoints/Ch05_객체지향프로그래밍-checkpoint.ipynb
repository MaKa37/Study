# 1. 클래스와 객체

```python
class Dog:
    # 생성자: 객체가 만들어질 때 자동으로 1회 호출됨
    def __init__(self, name, bread):
        self.name = name
        self.bread = bread

    # 인스턴스 메서드
    def bark(self):
        # self를 통해 내 이름에 접근
        print(f"{self.name}가 멍멍 짖습니다!")

# 객체 생성
dog1 = Dog("바둑이", "진돗개")
dog2 = Dog("초코", "푸들")

print(dog1.bread) # 진돗개
dog2.bark()       # 초코가 멍멍 짖습니다!
```

```python
class Car:
    # 1. 클래스 변수 (모든 객체가 공유)
    wheel_count = 4

    # 2. 생성자 및 3. 인스턴스 변수 (객체마다 고유함)
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
        self.speed = 0

    # 4. 인스턴스 메서드 (self 사용)
    def accelerate(self, value):
        self.speed += value
        print(f"{self.brand}가 {self.speed}km/h로 가속합니다.")

    # 5. 클래스 메서드(cls 사용)
    @classmethod
    def get_wheel_count(cls):
        return f"이 자동차의 기본 바퀴는 {cls.wheel_count}개 입니다."

    # 6. 정적 메서드 (self, cls 모두 사용 안 함)
    @staticmethod
    def check_license(age):
        return age >= 18

# 7. 객체 생성(인스턴스화)
my_car = Car("Hyundai", "Black")
your_car = Car("Kia", "White")

# 8. 객체 및 메서드 사용
my_car.accelerate(30)         # Hyundai가 30km/h로 가속합니다.
your_car.accelerate(50)       # Kia가 50km/h로 가속합니다.
print(Car.get_wheel_count())  # 이 자동차의 기본 바퀴는 4개 입니다.
print(Car.check_license(20))  # True
```

# 2. 객체지향의 핵심 메커니즘(확장과 유연성)

```python
# 다형성(Polymorphism)과 오버라이딩(Overriding)
class Animal:
    def speak(self): pass

class Cat(Animal):
    def speak(self):
        print("야옹")

class Cow(Animal):
    def speak(self):
        print("음메~")

# 다형성을 활용한 패턴: 동물의 종류를 몰라도 순회할 수 있음
animals = [Cat(), Cow(), Cat()]

for animal in animals:
    animal.speak()

# 야옹
# 음메~
# 야옹
```

```python
# 1. 부모 클래스 정의
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "알 수 없는 소리"

# 2. 자식 클래스들의 메서드 오버라이딩(Overriding)
class Dog(Animal):
    # 부모의 speak() 메서드를 개에 맞게 재정의
    def speak(self):
        return f"{self.name}가 멍멍 짖습니다!"

class Cat(Animal):
    # 부모의 speak() 메서드를 개에 맞게 재정의
    def speak(self):
        return f"{self.name}가 야옹 웁니다!"        

# 3. super() 활용 예시
class Bird(Animal):
    def __init__(self, name, can_fly):
        # 부모 클래스(Animal)의 생성자를 호출하여 name 초기화
        super().__init__(name)
        self.can_fly = can_fly

    def speak(self):
        return f"{self.name}가 짹짹 웁니다!"

# 4. 상속 관계가 없는 클래스 (덕 타이핑용)
class Robot:
    def speak(self):
        return "삐리삐리. 저는 로봇입니다."

# --- 다형성 (Polymorphism) 활용 --- #

# 상속받은 Dog, Cat, Bird 객체와 전혀 무관한 Robot 객체를 하나의 리스트로 묶음
entities = [
    Dog("바둑이"),
    Cat("나비"),
    Bird("참새", True),
    Robot()
]

for entity in entities:
    print(entity.speak())

"""
바둑이가 멍멍 짖습니다!
나비가 야옹 웁니다!
참새가 짹짹 웁니다!
삐리삐리. 저는 로봇입니다.
"""
```
