## 1.1 힙 큐(heapq)

```python
import heapq

# 1. 빈 리스트를 힙 처럼 사용
hq = []
heapq.heappush(hq, 4)
heapq.heappush(hq, 1)
heapq.heappush(hq, 7)
heapq.heappush(hq, 3)

print(hq) # 출력: [1, 3, 7, 4] -> 전체 정렬은 아니지만, 0번 인덱스(1)는 확실한 최솟값
print("-------------------------------------------------")

# 2. 기존 리스트 힙으로 변환 (O(N))
data = [5, 2, 8, 1, 9]
heapq.heapify(data)

print(data) # 출력: [1, 2, 8, 5, 9]
print(heapq.heappop(data)) # 출력: 1
print(heapq.heappop(data)) # 출력: 2
print(heapq.heappop(data)) # 출력: 5
print(heapq.heappop(data)) # 출력: 8
print(heapq.heappop(data)) # 출력: 9
print("-------------------------------------------------")

# 3. 최대 힙 구현
max_heap = []
values = [1, 5, 3, 2 ,4]

# 값을 푸시할 때 음수로 변환하여 넣음 (우선순위, 실제 값)
for val in values:
    heapq.heappush(max_heap, (-val, val))

print(max_heap) # 출력: [(-5, 5), (-4, 4), (-3, 3), (-1, 1), (-2, 2)]

# 팝을 할 때 실제 값을 가져옴 (가장 큰 값부터 나옴)
while max_heap:
    print(heapq.heappop(max_heap)[1]) # [-val, val] 중 인덱스 1번째인 val 출력: 5, 4, 3, 2, 1
print("-------------------------------------------------")

# 4. N Small list / N Large List
small_list = heapq.nsmallest(2, values)
large_list = heapq.nlargest(2, values)

print(small_list, large_list) # 출력: [1, 2] [5, 4] / values = [1, 5, 3, 2 ,4]
```

## 1.2 동기화 큐(queue.Queue)

```python
import queue
import threading

q = queue.Queue(maxsize=10) # 큐의 최대 크기 지정 가능
print(q) # 출력: <queue.Queue object at 0x000001E111A04610>
print(q.qsize())
print("-------------------------------------------------")

# 데이터 넣기 (큐가 꽉 차면 빈자리가 날 때까지 알아서 대기함)
q.put("작업1")
print(q.qsize())
print("-------------------------------------------------")

# 데이터 빼기 (큐가 비어있으면 데이터가 들어올 때까지 알아서 대기함)
item = q.get()
print(item)
print(q.qsize())
print(q.empty())
```

## 1.3 파이썬 원시 배열(array.array)

```python
import array
import sys

"""
- 'i' : 부호 있는 정수 (C의 `signed int`, 최소 2바이트)
- 'I' : 부호 없는 정수 (C의 `unsigned int`, 최소 2바이트)
- 'f' : 부동 소수점 (C의 `float`, 4바이트) 
- 'd' : 배정밀도 부동 소수점 (C의 `double`, 8바이트)
"""

# 일반 리스트 생성
py_list = [1, 2, 3, 4, 5]

# 'i'는 C언어의 signed int (보통 4 Bytes) 타입만 넣겠다는 타입 코드
arr = array.array('i', py_list)
print(arr)
print("-------------------------------------------------")

# 지정된 타입이 아닌 다른 값을 리스트에 넣을 시 에러 발생
# arr.append("EX") -> TypeError: integer argument expected, got String
print(f"List 메모리: {sys.getsizeof(py_list)} Bytes") # 출력: List 메모리: 104 Bytes
print(f"Array 메모리: {sys.getsizeof(arr)} Bytes")    # 출력: Array 메모리: 100 Bytes
```
