# 1. HTTP 프로토콜(Requests)

```python
import requests
import json

# 학습용 무료 REST API 엔드포인트
# (참고: 실제로 서버의 데이터가 영구적으로 바뀌지는 않지만, 성공/실패 상태 코드와 결과 포맷을 정확히 반환해 줍니다.)
BASE_URL = "https://jsonplaceholder.typicode.com/posts"

def print_result(method_name, response):
    """가독성을 높이기 위한 결과 출력용 헬퍼 함수"""
    print(f"[{method_name}] 상태 코드: {response.status_code}")
    # 응답 본문이 너무 길지 않게 들여쓰기하여 출력
    print(json.dumps(response.json(), indent=2, ensure_ascii=False))
    print("-" * 50)

# =====================================================================
# 1. GET (조회)
# 속성: 안전성(O), 멱등성(O)
# 용도: 데이터를 읽어올 때 사용. URL에 모든 정보가 포함됨.
# =====================================================================
# 1번 게시글 조회 요청
response_get = requests.get(f"{BASE_URL}/1")
print_result("GET", response_get)

# =====================================================================
# 2. POST (생성)
# 속성: 안전성(X), 멱등성(X)
# 용도: 새로운 데이터를 등록. 데이터는 URL이 아닌 본문(Body)에 담아 전송.
# =====================================================================
new_post_data = {
    "title": "크롤링 학습하기",
    "body": "HTTP 메서드 이해하기",
    "userId": 1
}
# requests.post의 'json' 파라미터를 사용하면 자동으로 헤더에 'Content-Type: application/json'이 추가됩니다.
response_post = requests.post(BASE_URL, json=new_post_data)
print_result("POST", response_post) # 성공 시 201 (Created) 상태 코드 반환

# =====================================================================
# 3. PUT (전체 수정)
# 속성: 안전성(X), 멱등성(O)
# 용도: 기존 자원을 완전히 새로운 데이터로 덮어쓰기. (여러 번 보내도 결과 동일)
# =====================================================================
put_update_data = {
    "id": 1,
    "title": "수정된 제목 (PUT)",
    "body": "내용이 완전히 교체되었습니다.",
    "userId": 1
}

# 1번 게시글 전체를 덮어씀
response_put = requests.put(f"{BASE_URL}/1", json=put_update_data)
print_result("PUT", response_put)

# =====================================================================
# 4. PATCH (일부 수정)
# 속성: 안전성(X), 멱등성(X/O)
# 용도: 자원의 일부분만 변경. (여기서는 title만 변경하므로 멱등성 유지)
# =====================================================================
patch_update_data = {
    "title": "제목만 살짝 수정(PATCH)"
    # body와 userId는 보내지 않으므로 기존 데이터가 유지됨
}

# 1번 게시글의 제목만 수정
response_patch = requests.patch(f"{BASE_URL}/1", json=patch_update_data)
print_result("PATCH", response_patch)

# =====================================================================
# 5. DELETE (삭제)
# 속성: 안전성(X), 멱등성(O)
# 용도: 특정 자원을 삭제. (이미 삭제된 것을 또 삭제 요청해도 상태는 '삭제됨'으로 동일)
# =====================================================================
# 1번 게시글 삭제 요청 (Body가 필요 없음)
response_delete = requests.delete(f"{BASE_URL}/1")
print_result("DELETE", response_delete)
```

```python
# -------------------------------------------------- #
# RESTful API & 파이썬 연동
# -------------------------------------------------- #
from fastapi import FastAPI
from fastapi.testclient import TestClient

# =====================================================================
# 1. 서버 측 구현 (RESTful API 라우팅 설계)
# =====================================================================
app = FastAPI()

# 가상의 데이터베이스 세팅
MOCK_USERS = [
    {"id": 123, "name": "김철수", "role": "admin"},
    {"id": 124, "name": "이영희", "role": "user"},
]
MOCK_ORDERS = [
    {"id": 45, "user_id": 123, "item": "노트북"},
    {"id": 46, "user_id": 124, "item": "무선 마우스"},
]

# [1] 경로 변수 (Path Variable)
@app.get("/users/{user_id}")
def get_user(user_id: int):
    # 전달받은 user_id로 자원 식별
    user = next((u for u in MOCK_USERS if u["id"] == user_id), None)
    return {"status": "success", "data": user}

# [2] 계층적 경로 (Hierarchical Path)
@app.get("/users/{user_id}/orders/{order_id}")
def get_user_order(user_id: int, order_id: int):
    # 포함 관계(특정 유저의 특정 주문) 자원 식별
    order = next((o for o in MOCK_ORDERS if o["user_id"] == user_id and o["id"] == order_id), None)
    return {"status": "success", "data": order}

# [3 & 4] 쿼리 파라미터, 페이징 및 복합 쿼리 (Query Parameter, Pagination)
@app.get("/users")
def get_users(role: str = None, page: int = 1, limit: int = 10):
    # 1. 쿼리 파라미터 필터링
    filtered_users = MOCK_USERS
    if role:
        filtered_users = [u for u in filtered_users if u["role"] == role]
    
    # 2. 페이징 로직 (테스트를 위해 가상 데이터를 임의로 복제하여 양을 늘림)
    extended_users = filtered_users * 50  
    
    start_idx = (page - 1) * limit
    end_idx = start_idx + limit
    paginated_users = extended_users[start_idx:end_idx]

    return {
        "status": "success",
        "meta": {"role_filter": role, "page": page, "limit": limit, "returned_count": len(paginated_users)},
        "data": paginated_users
    }


# =====================================================================
# 2. 클라이언트 측 호출 (requests 구현 시뮬레이션)
# =====================================================================
# TestClient는 파이썬 내부에서 requests 라이브러리와 100% 동일한 인터페이스를 제공합니다.
client = TestClient(app)

def run_simulation():
    print("--- [1] 경로 변수 (Path Variable) ---")
    user_id = 123
    response1 = client.get(f"/users/{user_id}")
    print(f"생성된 URL : {response1.request.url}")
    print(f"응답 결과  : {response1.json()}\n")

    print("--- [2] 계층적 경로 (Hierarchical Path) ---")
    order_id = 45
    response2 = client.get(f"/users/{user_id}/orders/{order_id}")
    print(f"생성된 URL : {response2.request.url}")
    print(f"응답 결과  : {response2.json()}\n")

    print("--- [3] 쿼리 파라미터 (Query Parameter) ---")
    response3 = client.get("/users", params={"role": "admin"})
    print(f"생성된 URL : {response3.request.url}")
    print(f"응답 결과  : {response3.json()['meta']}\n") # 출력 길이를 위해 meta만 출력

    print("--- [4] 페이징 및 복합 쿼리 (Pagination & Complex) ---")
    params = {"role": "admin", "page": 2, "limit": 3}
    response4 = client.get("/users", params=params)
    print(f"생성된 URL : {response4.request.url}")
    print(f"응답 메타  : {response4.json()['meta']}")
    print(f"응답 데이터: {response4.json()['data']}")

if __name__ == "__main__":
    run_simulation()
```

```python
import requests
import time

url = "https://jsonplaceholder.typicode.com/posts"
page = 1
all_posts = []

while True:
    # [학습 포인트 1] 쿼리 파라미터 자동화 (REST 아키텍처 활용)
    # params 딕셔너리를 넘기면 requests가 자동으로 URL 뒤에 '?page=1&limit=100' 형태로 조립합니다.
    # URL 문자열을 직접 수정할 필요가 없어 가독성과 안정성이 매우 높아집니다.
    response = requests.get(url, params={"_page": page, "limit": 20})
    
    # [학습 포인트 2] 선제적 에러 방어
    # 200번대(성공) 응답이 아닐 경우 즉시 예외(HTTPError)를 발생시킵니다.
    # 404, 429, 500 에러 발생 시 쓰레기 데이터가 리스트에 섞이는 것을 막는 훌륭한 안전장치입니다.
    response.raise_for_status() 
    
    # 응답받은 JSON 문자열을 파이썬 리스트/딕셔너리 객체로 변환
    data = response.json()
    
    # [학습 포인트 3] 루프 종료 조건 (안전한 탈출)
    # 서버가 빈 리스트([])를 주거나 더 이상 반환할 데이터가 없으면 무한 루프를 빠져나옵니다.
    if not data:
        break
        
    # [학습 포인트 4] 데이터 병합 (append vs extend)
    # append를 쓰면 [[1,2], [3,4]] 형태로 중첩되지만, 
    # extend를 쓰면 [1,2,3,4] 형태로 평탄하게 합쳐집니다. 대용량 데이터 병합에 필수적입니다.
    all_posts.extend(data)
    
    # 다음 페이지 조회를 위해 변수 업데이트
    page += 1
    
    # [학습 포인트 5] 크롤링 매너 (IP 차단 방어)
    # 서버에 기계적인 과부하를 주지 않기 위해 루프 1회당 0.5초씩 의도적으로 대기합니다.
    time.sleep(0.5) 

print(f"총 {len(all_posts)}개의 데이터를 성공적으로 수집했습니다.")
for post in all_posts[:5]:
    print(f"Post ID: {post['id']}ㅣPost title: {post['title']}")
```

```python
import os
import time
import requests
from requests.exceptions import RequestException, Timeout, SSLError
from dotenv import load_dotenv

# [학습 포인트 1] 환경변수 로드 (.env 파일에서 API KEY 등을 안전하게 불러옴)
load_dotenv()

def fetch_data_safely(url, max_retries=3):
    """
    주어진 URL에서 데이터를 안전하게 수집하고 예외를 처리하는 함수입니다.
    """
    # 순수 URL 문자열 및 요청 헤더 설정
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/120.0.0.0 Safari/537.36",
        "Authorization": f"Bearer {os.getenv('MY_API_TOKEN', 'default_token')}"
    }
    
    for attempt in range(max_retries):
        try:
            # [학습 포인트 2] 타임아웃 튜플 분리 (연결 타임아웃 3초, 읽기 타임아웃 10초)
            response = requests.get(url, headers=headers, timeout=(3, 10))
            
            # [학습 포인트 3] 429 Too Many Requests 방어 및 재시도 로직
            if response.status_code == 429:
                retry_after = int(response.headers.get("Retry-After", 60))
                print(f"⚠️ Rate Limit 초과. {retry_after}초 대기 후 재시도 ({attempt + 1}/{max_retries})")
                time.sleep(retry_after)
                continue 
                
            # 200번대 성공 응답인지 체크 (4xx, 5xx 에러면 여기서 예외 발생)
            response.raise_for_status() 
            
            # [학습 포인트 4] 데이터 파싱 및 타입 에러 방어
            # json() 내부적으로 인코딩 처리를 하므로 apparent_encoding 불필요
            try:
                data = response.json()
                print(f"✅ 조회 성공! 제목: {data.get('title', '제목 없음')}")
                return data  # 성공 시 데이터 반환 및 함수 완전 종료
                
            except requests.exceptions.JSONDecodeError:
                print("❌ 서버가 JSON을 반환하지 않았습니다. (HTML 에러 페이지 또는 방화벽 차단 의심)")
                print(f"응답 본문 확인:\n{response.text[:200]}")
                break  # 파싱 불가능한 데이터이므로 재시도하지 않고 루프 탈출
                
        # [학습 포인트 5] 기업망/공공기관 데이터 수집의 복병: SSL 에러 방어
        except SSLError:
            print("🔒 SSL 인증서 에러. (임시방편: verify=False 옵션 고려. 단 보안 주의)")
            break
            
        # [학습 포인트 6] 타임아웃과 일반 네트워크 에러를 명시적으로 분리하여 대응 차별화
        except Timeout:
            print(f"⏳ 서버 응답 지연 타임아웃 발생. 5초 대기 후 재시도 ({attempt + 1}/{max_retries})")
            time.sleep(5)
            
        # [학습 포인트 7] RequestException은 모든 requests 예외의 최상위 클래스이므로 가장 마지막에 배치
        except RequestException as e:
            print(f"🚨 네트워크 요청 실패 또는 API 에러: {e}")
            time.sleep(2)
            
    # [학습 포인트 8] 파이썬 for-else 문법
    # break로 루프를 탈출하지 못하고 재시도 횟수를 모두 소진했을 때만 실행됨
    else:
        print(f"❌ 최대 재시도 횟수({max_retries}회)를 초과하여 데이터 수집에 실패했습니다. URL: {url}")
        return None

# =====================================================================
# 실행 테스트 구간
# =====================================================================
if __name__ == "__main__":
    target_url = "https://jsonplaceholder.typicode.com/posts/1"
    fetch_data_safely(target_url)
```

```python
import time
import requests

def run_session_example():
    """
    requests.Session()을 활용한 성능 최적화 및 상태 유지 학습용 예제 코드
    """
    url_1 = "https://jsonplaceholder.typicode.com/posts/1"
    url_2 = "https://jsonplaceholder.typicode.com/posts/2"

    print("--- Session 객체를 활용한 데이터 통신 시작 ---")
    
    # [학습 포인트 1] 메모리 누수 방지를 위한 안전한 세션 관리
    # with 문을 활용하면 크롤링/통신 작업 종료 후 session.close()가 자동 호출되어 
    # 시스템 자원(소켓 등)을 안전하게 반환합니다.
    with requests.Session() as session:

        # [학습 포인트 2] 공통 헤더 단일 등록 (코드 중복 제거)
        # 세션 객체에 헤더를 한 번만 업데이트해 두면, 이후 이 세션에서 파생되는 
        # 모든 get/post 요청에 해당 헤더가 자동으로 삽입됩니다.
        session.headers.update({
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/120.0.0.0 Safari/537.36",
            "Authorization": "Bearer MY_SECRET_TOKEN"
        })

        # [학습 포인트 3] 상태(쿠키) 자동 유지 (로그인 기반 크롤링 시 필수)
        # 예시: session.post("https://example.com/login", data={"id": "myid", "pw": "mypw"})
        # 위와 같이 로그인 요청을 한 번 성공하면 서버가 발급한 세션 쿠키가 session 객체 내부에 저장되며,
        # 이후 요청부터는 로그인 쿠키가 자동으로 묻어갑니다.

        # [학습 포인트 4] TCP 연결 유지(Keep-Alive)로 인한 속도 비약적 향상
        # 단발성 requests.get()은 매번 연결을 새로 맺고 끊지만, 
        # Session은 한 번 맺은 연결 파이프(TCP Socket)를 재사용하므로 훨씬 빠릅니다.
        start_time = time.time()

        # 동일한 도메인(jsonplaceholder.typicode.com)에 여러 번 요청할 때 그 진가가 드러납니다.
        response1 = session.get(url_1, timeout=5)
        response2 = session.get(url_2, timeout=5)

        end_time = time.time()

        # 결과 출력
        print(f"✅ 첫 번째 데이터 조회 성공: {response1.json().get('title')}")
        print(f"✅ 두 번째 데이터 조회 성공: {response2.json().get('title')}")
        print(f"⚡ 세션 기반 연속 통신 소요 시간: {end_time - start_time:.4f}초")

if __name__ == "__main__":
    run_session_example()
```
