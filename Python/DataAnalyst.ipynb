```python
import pandas as pd
import numpy as np
```

```python
# Numpy 연산
x = np.array([10, 20, 30])
print(x)
type(x)

y = x / 7
print(y, "\n")

x = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 6.0])

print(x + y)
print(x - y)
print(x * y)
print(x / y, "\n")

print("### N차 배열 ###")
A = np.array([[10, 11],[20, 22]])
print(A, "\n")
B = np.array([[3, 6],[5, 7]])
print(B, "\n")

print("### N차 배열 합연산 ###")
print(A + B, "\n")

print("### N차 배열 곱연산 ###")
print(A * B, "\n")

print("### N차 배열 합연산 ###")
print(A + A + A, "\n")
```

```python
import matplotlib.pyplot as plt
```

```python
x = np.arange(0, 10, 0.1)
y = np.sin(x)
plt.plot(x, y)
plt.show()
```

```python
x = np.arange(0, 10, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x, y1, label = 'sin')
plt.plot(x, y2, label = 'cos', linestyle='--')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('sin & cos')
plt.legend()
plt.show()
```

```python
D2 = eData1
D2["sex"] = D2["sex"].astype("category")
D2["edu"] = D2["edu"].astype("category")
print(D2.head(3))

print()
print(D2.groupby("sex")["salary"].describe())

print()
print(D2.groupby("edu")["salary"].describe())
```

```python
D2_Group_by_sex = D2.groupby("sex")["salary"]
print(D2_Group_by_sex.mean())
print(D2_Group_by_sex.std())
```

```python
#빈도표 및 분할표 구하기
sex_freq = pd.crosstab(index=D2["sex"], columns="count")
print(sex_freq)
print()

edu_freq = pd.crosstab(index=D2["edu"], columns="count")
print(edu_freq)
print()

sex_edu_table = pd.crosstab(index=D2["sex"], columns=D2["edu"])
print(sex_edu_table)
```

```python
plt.bar(sex_freq.index, sex_freq["count"])
```
