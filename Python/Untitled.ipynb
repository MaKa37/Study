```python
from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])  # 큐 초기화
    order = []

    while queue:
        node = queue.popleft()  # 맨 앞에서 꺼냄 (FIFO)
        order.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)  # 맨 뒤에 추가

    return order

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C']
}

print(bfs(graph, 'A'))  # ['A', 'B', 'C', 'D']
```
